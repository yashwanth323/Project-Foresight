# FORESIGHT | Exploratory Data Analysis
Explore daily demand, seasonality, product mix, missing values and outliers before modelling.

```python
import pandas as pd
import plotly.express as px
df = pd.read_csv('../data/processed/sales_clean.csv', parse_dates=['date'])
px.line(df.groupby('date',as_index=False).quantity_sold.sum(), x='date', y='quantity_sold', title='Total daily demand')
```
